```python
from langgraph.checkpoint.memory import MemorySaver
from langgraph.graph import StateGraph, START, END
```

```python
# 1. Initialize the Checkpointer (The 'Memory Card')
memory = MemorySaver()

# 2. Define a simple graph
builder = StateGraph(dict)
builder.add_node("step_1", lambda x: {"count": 1})
builder.add_edge(START, "step_1")
builder.add_edge("step_1", END)
```

```python
# 3. Compile with the checkpointer!
agent = builder.compile(checkpointer=memory)

# 4. Use a unique Thread ID for the user
config = {"configurable": {"thread_id": "user_4321"}}
```

```python
# Run the agent
agent.invoke({"input": "hello"}, config)
```

```python
# Later... the agent can retrieve this state using the same ID!
state = agent.get_state(config)
print(f"Saved Count: {state.values['count']}")
```
